```python
import pandas as pd

df = pd.read_csv("../data/interim/Net_renewables_generation.csv", skiprows=4)

# Keep only the rows you want
df_clean = df[df["description"].str.contains(r": all commercial|: all industrial", na=False)].copy()

# Split description into two join fields
df_clean[["state_or_region", "sector"]] = df_clean["description"].str.split(" : ", expand=True)

# Make sector names match the prices file
df_clean["sector"] = df_clean["sector"].str.replace("all ", "", regex=False)

# Rename source key for convenience
df_clean = df_clean.rename(columns={"source key": "source_key"})

# Drop regions/national rows if your prices file is state-only
regions_to_drop = [
    "United States",
    "New England",
    "Middle Atlantic",
    "East North Central",
    "West North Central",
    "South Atlantic",
    "East South Central",
    "West South Central",
    "Mountain",
    "Pacific Contiguous",
    "Pacific Noncontiguous"
]

df_clean = df_clean[~df_clean["state_or_region"].isin(regions_to_drop)].copy()

# Find year columns
year_cols = [c for c in df_clean.columns if c.isdigit()]


# Reshape to long
renewables_long = df_clean.melt(
    id_vars=["state_or_region", "sector", "units", "source_key"],
    value_vars=year_cols,
    var_name="year",
    value_name="renewable_generation"
)

# Clean values
renewables_long["renewable_generation"] = renewables_long["renewable_generation"].replace("--", pd.NA)
renewables_long["renewable_generation"] = pd.to_numeric(
    renewables_long["renewable_generation"], errors="coerce"
)

# Make sure year is numeric
renewables_long["year"] = renewables_long["year"].astype(int)

# Drop empty rows
renewables_long = renewables_long.dropna(subset=["renewable_generation"]).copy()

# Final column order
renewables_long = renewables_long[
    ["state_or_region", "sector", "year", "renewable_generation", "units", "source_key"]
].sort_values(["state_or_region", "sector", "year"]).reset_index(drop=True)

print(renewables_long.head())
print(renewables_long.shape)
print(renewables_long.dtypes)
```

```python
renewables_long.to_csv("../data/processed/renewables_clean.csv", index=False)
```
